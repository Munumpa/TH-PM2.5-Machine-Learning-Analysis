```python
from google.colab import drive
import pdfplumber  # เครื่องมือสำหรับสกัดตารางออกจากไฟล์ PDF
import pandas as pd
import os
import glob

# เชื่อมต่อกับ Google Drive เพื่อเข้าถึงโฟลเดอร์ที่เก็บไฟล์ข้อมูลดิบ
drive.mount('/content/drive')

# กำหนดเส้นทาง (Path) ไปยังโฟลเดอร์ที่เก็บไฟล์ PDF รายงานฝุ่น PM2.5 ทั้งหมด
folder_path = '/content/drive/MyDrive/Data_Sci_Pro/Data_Set/PM25/PM25_Data'

# ใช้ glob เพื่อกวาดหาไฟล์ที่มีนามสกุล .pdf ทั้งหมดในโฟลเดอร์นั้น
pdf_files = glob.glob(os.path.join(folder_path, '*.pdf'))

all_data = [] # สร้าง List ว่างไว้รอรับข้อมูลตารางจากทุกไฟล์

# วนลูปเข้าไปอ่านไฟล์ PDF ทีละไฟล์
for file in pdf_files:
    print(f"กำลังประมวลผลไฟล์: {os.path.basename(file)}")

    # เปิดไฟล์ PDF ด้วย pdfplumber
    with pdfplumber.open(file) as pdf:
        # วนลูปอ่านข้อมูลทีละหน้า (Page) ในไฟล์นั้นๆ
        for page in pdf.pages:
            # คำสั่งดึงข้อมูลที่หน้าตาเหมือน "ตาราง" ออกมา
            tables = page.extract_tables()

            # ใน 1 หน้าอาจมีหลายตาราง จึงต้องวนลูปอีกครั้ง
            for table in tables:
                # แปลงข้อมูลตารางที่ดึงมาได้ ให้กลายเป็น DataFrame ของ Pandas
                df_temp = pd.DataFrame(table)

                # สร้างคอลัมน์ใหม่เพื่อเก็บ "ชื่อไฟล์ต้นฉบับ" ไว้
                # เพราะชื่อไฟล์มักจะซ่อนข้อมูลสำคัญไว้ เช่น รหัสสถานี และ ปี (เช่น 11T_2020.pdf)
                # ซึ่งต้องเอาไปใช้แยกแยะข้อมูลในขั้นตอนต่อไป
                df_temp['Source_File'] = os.path.basename(file)

                # เก็บตารางที่ได้เข้า List รวม
                all_data.append(df_temp)

if all_data:
    # นำตารางจากทุกหน้า ทุกไฟล์ มาต่อกันเป็นตารางยาวๆ ตารางเดียว
    final_df = pd.concat(all_data, ignore_index=True)

    # Data Cleaning เบื้องต้น: ลบแถวที่ว่างเปล่า (ไม่มีข้อมูลเลยสักคอลัมน์) ทิ้งไป
    final_df = final_df.dropna(how='all')

    # บันทึกผลลัพธ์เป็นไฟล์ CSV (ข้อมูลที่ได้จะยังดูยุ่งเหยิงอยู่ ซึ่งเราจะคลีนต่อในโค้ดบล็อกถัดไป)
    final_df.to_csv('Extracted_PM25_Data.csv', index=False, encoding='utf-8-sig')

    display(final_df.head(10))
else:
    print("ไม่พบตารางในไฟล์ PDF")
```

```python
  import pandas as pd
  import numpy as np
  import re

  # โหลดข้อมูลดิบที่ได้จากการสกัด PDF
  df = pd.read_csv('Extracted_PM25_Data.csv')
  results = []


  # สร้าง Dictionary แก้ปัญหาตัวอักษรเพี้ยน
  # ปัญหา: ข้อมูลที่สกัดจาก PDF ภาษาไทยมักมีปัญหา "สระจม/สระลอย/พิมพ์ผิด"
  # วิธีแก้: ดักจับ keyword ที่เป็นไปได้ทั้งหมด แล้ว Map ให้เป็นชื่อเดือนภาษาอังกฤษมาตรฐาน
  thai_to_eng_months = {
      'มก': 'Jan', 'กุม': 'Feb', 'กมุ': 'Feb', 'มีน': 'Mar', 'เมษ': 'Apr',
      'พฤษ': 'May', 'มิถ': 'Jun', 'กรก': 'Jul', 'สิง': 'Aug', 'สงิ': 'Aug',
      'กัน': 'Sep', 'กนั': 'Sep', 'ตุล': 'Oct', 'ตลุ': 'Oct', 'พฤศ': 'Nov',
      'ธัน': 'Dec', 'ธนั': 'Dec'
  }

  # วนลูปจัดการข้อมูลทีละไฟล์ (จัดกลุ่มตามชื่อไฟล์ PDF ต้นฉบับ)
  for file_name, group in df.groupby('Source_File', sort=False):

      # กรองเอาเฉพาะบรรทัด (Row) ที่มีคำใบ้ว่าเป็นชื่อเดือนอยู่ในคอลัมน์แรก ('0')
      month_rows = group[group['0'].astype(str).str.contains('|'.join(thai_to_eng_months.keys()), na=False)]
      if month_rows.empty: continue

      # กำหนดขอบเขตของ "หัวตาราง (Header)" เพื่อนำไปค้นหาคอลัมน์เป้าหมาย
      first_idx = month_rows.index[0]
      header_block = group.loc[:first_idx-1]

      # ค้นหาคอลัมน์เป้าหมายแบบไดนามิก (Dynamic Column Finder)
      # ปัญหา: ตารางในแต่ละปีหรือแต่ละสถานี อาจมีจำนวนคอลัมน์ไม่เท่ากัน (ตำแหน่งฝุ่น PM2.5 ไม่อยู่ที่เดิม)
      # วิธีแก้: เขียนโค้ดกวาดหาคอลัมน์ที่มีคำว่า '2.5' และ 'รายเดือน' (พร้อมดักจับคำที่อาจจะสะกดเพี้ยน)
      target_col = None
      for col in group.columns[:-1]:
          # ยุบรวมข้อความหัวตารางที่อาจจะถูกตัดขึ้นบรรทัดใหม่ (\n) ให้อยู่ในบรรทัดเดียว
          text = " ".join(header_block[col].dropna().astype(str)).replace('\n', ' ')

          # ค้นหาเงื่อนไข: ต้องมีคำว่า 2.5 และมีคำว่า รายเดือน
          if '2.5' in text and any(kw in text for kw in ['รายเดือน', 'รายเดอื น', 'รายเดอน', 'รายเดือ']):
              target_col = col
              break

      # กรณีที่หัวตารางหลัก (PM2.5) กับหัวตารางย่อย (รายเดือน) ถูกแยกอยู่คนละคอลัมน์
      if target_col is None:
          current_main_header = ""
          for col in group.columns[:-1]:
              col_text = " ".join(header_block[col].dropna().astype(str)).replace('\n', ' ')
              if '2.5' in col_text: current_main_header = "PM2.5" # จำไว้ว่าเจอหมวด PM2.5 แล้ว

              if current_main_header == "PM2.5" and any(kw in col_text for kw in ['รายเดือน', 'รายเดอื น', 'รายเดอน', 'รายเดือ']):
                  target_col = col
                  break

      if target_col is None: continue # ถ้าหาไม่เจอจริงๆ ให้ข้ามไฟล์นี้ไป

      # สกัดข้อมูล (Extraction) และทำความสะอาด (Cleaning)
      # ดึงชื่อ "รหัสสถานี" และ "ปี" ออกมาจากชื่อไฟล์ด้วย Regular Expression (Regex)
      # เช่น ไฟล์ชื่อ 80T_2020.pdf จะได้ station='80T', year='2020'
      match = re.search(r'(\w+)_(\d{4})', file_name)
      station = match.group(1) if match else file_name
      year = match.group(2) if match else 'Unknown'

      # เริ่มดึงค่าฝุ่นในแต่ละเดือน
      for idx, row in month_rows.iterrows():
          month_cell = str(row['0'])
          val_cell = str(row[target_col]) if pd.notna(row[target_col]) else ""

          # ปัญหาใหญ่: PDF บางตาราง ข้อมูลหลายเดือนถูกยุบรวม (Merge) อยู่ในเซลล์เดียวกันคั่นด้วย '\n'
          # วิธีแก้: แตกข้อมูล (Split) ออกมาเป็น List แล้วจับคู่เดือนกับค่าฝุ่นทีละตัว
          if '\n' in month_cell and len(month_cell.split('\n')) >= 6:
              m_list = month_cell.split('\n')
              v_list = val_cell.split('\n')
              for m_raw, v_raw in zip(m_list, v_list):
                  m_clean = next((eng for thai, eng in thai_to_eng_months.items() if thai in m_raw), None)
                  if m_clean is None: continue

                  # แปลงค่าฝุ่นเป็นตัวเลข (Float) ถ้าแปลงไม่ได้ให้ใส่เป็นค่าว่าง (NaN) ไว้ก่อน
                  try: val = float(v_raw.strip())
                  except: val = np.nan
                  results.append({'Station': station, 'Year': year, 'Month': m_clean, 'PM2.5': val})

          # กรณีปกติ: 1 เดือน ต่อ 1 บรรทัดตาราง
          else:
              m_clean = next((eng for thai, eng in thai_to_eng_months.items() if thai in month_cell), None)
              if m_clean is None: continue
              try: val = float(val_cell.strip())
              except: val = np.nan
              results.append({'Station': station, 'Year': year, 'Month': m_clean, 'PM2.5': val})


  # จัดเก็บผลลัพธ์
  # แปลงข้อมูลทั้งหมดกลับเป็น DataFrame รูปแบบมาตรฐาน (โครงสร้าง: สถานี, ปี, เดือน, ค่าฝุ่น)
  parsed_df = pd.DataFrame(results)
  parsed_df.to_csv('Cleaned_PM25_Monthly.csv', index=False)
  print(f"จำนวนข้อมูลทั้งหมด: {len(parsed_df)} แถว")
  display(parsed_df.head(10))
```

```python
import pandas as pd

# 1. โหลดข้อมูล PM2.5
df_pm = pd.read_csv('Cleaned_PM25.csv')

# 2. เติมค่าว่าง (NaN) ด้วยค่าเฉลี่ยของเดือนนั้นๆ ในจังหวัดเดียวกัน
# ใช้ transform เพื่อกระจายค่าเฉลี่ยกลับลงไปในแถวที่ว่าง
monthly_avg = df_pm.groupby(['Province', 'Month'])['PM2.5'].transform('mean')
df_pm['PM2.5'] = df_pm['PM2.5'].fillna(monthly_avg)

# 3. กรณีที่ยังว่างอยู่ (เพราะจังหวัดนั้นไม่มีข้อมูลเดือนนั้นเลยทุกปี)
# ให้ใช้ค่าเฉลี่ยของเดือนนั้นๆ ทั้งประเทศมาแทน
national_monthly_avg = df_pm.groupby('Month')['PM2.5'].transform('mean')
df_pm['PM2.5'] = df_pm['PM2.5'].fillna(national_monthly_avg)

# 4. บันทึกผลลัพธ์
df_pm.to_csv('Cleaned_PM25_MonthlyV3.csv', index=False, encoding='utf-8-sig')
print(f"รวมข้อมูลเหลือ {len(df_pm)} แถว")
display(df_pm.head(10))
```

```python
import pandas as pd
import numpy as np

# 1. โหลดข้อมูลเดิม
df_pm25 = pd.read_csv('Cleaned_PM25_MonthlyV3.csv')

# 2. สร้างโครงสร้างข้อมูลที่สมบูรณ์ (77 จังหวัด x 12 เดือน x 5 ปี)
years = [2020, 2021, 2022, 2023, 2024]
months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun', 'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec']
# รายชื่อ 77 จังหวัด
provinces_77 = [
    # กรุงเทพมหานครและปริมณฑล
    'กรุงเทพมหานคร', 'นนทบุรี', 'ปทุมธานี', 'สมุทรปราการ', 'สมุทรสาคร', 'นครปฐม',

    # ภาคกลาง
    'พระนครศรีอยุธยา', 'สระบุรี', 'ลพบุรี', 'สิงห์บุรี', 'อ่างทอง', 'ชัยนาท', 'สมุทรสงคราม', 'นครสวรรค์', 'อุทัยธานี',

    # ภาคเหนือ
    'เชียงราย', 'เชียงใหม่', 'น่าน', 'พะเยา', 'พิษณุโลก', 'แพร่', 'แม่ฮ่องสอน', 'ลำปาง', 'ลำพูน', 'ตาก', 'อุตรดิตถ์', 'สุโขทัย', 'พิจิตร', 'เพชรบูรณ์', 'กำแพงเพชร',

    # ภาคตะวันออกเฉียงเหนือ (อีสาน)
    'ขอนแก่น', 'นครราชสีมา', 'นครพนม', 'เลย', 'หนองคาย', 'อุบลราชธานี', 'สกลนคร', 'อุดรธานี', 'บุรีรัมย์', 'มุกดาหาร', 'บึงกาฬ', 'กาฬสินธุ์', 'ชัยภูมิ', 'ยโสธร', 'อำนาจเจริญ', 'สุรินทร์', 'ศรีสะเกษ', 'ร้อยเอ็ด', 'มหาสารคาม', 'หนองบัวลำภู',

    # ภาคตะวันออก
    'ฉะเชิงเทรา', 'ชลบุรี', 'ตราด', 'ระยอง', 'สระแก้ว', 'ปราจีนบุรี', 'จันทบุรี', 'นครนายก',

    # ภาคตะวันตก
    'กาญจนบุรี', 'ราชบุรี', 'สุพรรณบุรี', 'ประจวบคีรีขันธ์', 'เพชรบุรี',

    # ภาคใต้
    'นราธิวาส', 'นครศรีธรรมราช', 'ภูเก็ต', 'ยะลา', 'สงขลา', 'สุราษฎร์ธานี', 'สตูล', 'ตรัง', 'ชุมพร', 'กระบี่', 'พัทลุง', 'ปัตตานี', 'ระนอง', 'พังงา'
] # ใส่ให้ครบ 77

full_grid = pd.MultiIndex.from_product([years, months, provinces_77], names=['Year', 'Month', 'Province'])
df_complete = pd.DataFrame(index=full_grid).reset_index()

# 3. นำข้อมูลที่มีอยู่มาใส่ในโครงสร้างใหม่
df_merged = pd.merge(df_complete, df_pm25, on=['Year', 'Month', 'Province'], how='left')

# 4. เติมค่าที่หายไป (Imputation)
# วิธีที่ 1: เติมด้วยค่าเฉลี่ยรายเดือนของจังหวัดนั้น (ถ้าปีอื่นมีข้อมูล)
df_merged['PM2.5'] = df_merged.groupby(['Province', 'Month'])['PM2.5'].transform(lambda x: x.fillna(x.mean()))

# วิธีที่ 2: ถ้าจังหวัดนั้นไม่มีสถานีเลย (หายไปทั้งจังหวัด) ให้ใช้ค่าเฉลี่ยของ "ภูมิภาค" หรือจังหวัดใกล้เคียงแทน
# (ตัวอย่างนี้ใช้ค่าเฉลี่ยรวมทั้งประเทศในเดือนนั้นๆ เพื่อไม่ให้โมเดลพัง)
df_merged['PM2.5'] = df_merged['PM2.5'].fillna(df_merged.groupby('Month')['PM2.5'].transform('mean'))

# 5. ตรวจสอบจำนวนแถว
print(f"จำนวนแถวที่ได้: {len(df_merged)} แถว") # จะต้องได้ 4620 พอดี
df_merged.to_csv('Cleaned_PM25_MonthlyV4.csv', index=False, encoding='utf-8-sig')
display(df_merged.head(10))
```
